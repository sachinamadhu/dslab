```python
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# a) Load the Iris dataset and display first five records
# ---------------------------------------------------------

df = pd.read_csv("/Iris - Iris.csv")

print("First five records:")
print(df.head())


# ---------------------------------------------------------
# b) Scatter plot: sepal.length vs sepal.width
# ---------------------------------------------------------

plt.figure(figsize=(7, 5))
plt.scatter(df["sepal.length"], df["sepal.width"], color="blue")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Sepal Length vs Sepal Width")
plt.grid(True)
plt.show()


# ---------------------------------------------------------
# c) Scatter plot with different colors for each variety
# ---------------------------------------------------------

colors = {
    "Setosa": "red",
    "Versicolor": "green",
    "Virginica": "blue"
}

plt.figure(figsize=(7, 5))

for variety, color in colors.items():
    data = df[df["variety"] == variety]
    plt.scatter(
        data["sepal.length"],
        data["sepal.width"],
        color=color,
        label=variety
    )

plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Sepal Length vs Sepal Width by Iris Variety")
plt.legend()
plt.grid(True)
plt.show()


# ---------------------------------------------------------
# d) Line plot: sepal.length vs sepal.width
# ---------------------------------------------------------

plt.figure(figsize=(8, 5))
plt.plot(
    df["sepal.length"],
    df["sepal.width"],
    color="purple",
    marker="o",
    linestyle="-"
)
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Line Plot of Sepal Length vs Sepal Width")
plt.grid(True)
plt.show()


# ---------------------------------------------------------
# e) Histogram of sepal.length
# ---------------------------------------------------------

plt.figure(figsize=(7, 5))
plt.hist(
    df["sepal.length"],
    bins=10,
    color="skyblue",
    edgecolor="black"
)
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.title("Distribution of Sepal Length")
plt.grid(axis="y")
plt.show()


# ---------------------------------------------------------
# f) Bar chart: total sepal.length for each variety
# ---------------------------------------------------------

total_sepal_length = df.groupby("variety")["sepal.length"].sum()

plt.figure(figsize=(7, 5))
plt.bar(
    total_sepal_length.index,
    total_sepal_length.values,
    color=["red", "green", "blue"]
)
plt.xlabel("Iris Variety")
plt.ylabel("Total Sepal Length")
plt.title("Total Sepal Length by Iris Variety")
plt.show()


# ---------------------------------------------------------
# g) Box plot: sepal.length for different varieties
# ---------------------------------------------------------

setosa = df[df["variety"] == "Setosa"]["sepal.length"]
versicolor = df[df["variety"] == "Versicolor"]["sepal.length"]
virginica = df[df["variety"] == "Virginica"]["sepal.length"]

plt.figure(figsize=(7, 5))
plt.boxplot(
    [setosa, versicolor, virginica],
    labels=["Setosa", "Versicolor", "Virginica"]
)
plt.xlabel("Iris Variety")
plt.ylabel("Sepal Length")
plt.title("Distribution of Sepal Length by Iris Variety")
plt.grid(axis="y")
plt.show()


# ---------------------------------------------------------
# h) Pie chart: percentage distribution of Iris varieties
# ---------------------------------------------------------

variety_counts = df["variety"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    variety_counts.values,
    labels=variety_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    colors=["red", "green", "blue"]
)
plt.title("Percentage Distribution of Iris Varieties")
plt.show()


# ---------------------------------------------------------
# i) Scatter plot: petal.length vs petal.width
#    with different colors for each variety
# ---------------------------------------------------------

plt.figure(figsize=(7, 5))

for variety, color in colors.items():
    data = df[df["variety"] == variety]
    plt.scatter(
        data["petal.length"],
        data["petal.width"],
        color=color,
        label=variety
    )

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Petal Length vs Petal Width")
plt.legend()
plt.grid(True)
plt.show()


# ---------------------------------------------------------
# j) Multiple subplots containing four histograms
# ---------------------------------------------------------

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Sepal Length
axes[0, 0].hist(
    df["sepal.length"],
    bins=10,
    color="skyblue",
    edgecolor="black"
)
axes[0, 0].set_title("Sepal Length")
axes[0, 0].set_xlabel("Length")
axes[0, 0].set_ylabel("Frequency")

# Sepal Width
axes[0, 1].hist(
    df["sepal.width"],
    bins=10,
    color="lightgreen",
    edgecolor="black"
)
axes[0, 1].set_title("Sepal Width")
axes[0, 1].set_xlabel("Width")
axes[0, 1].set_ylabel("Frequency")

# Petal Length
axes[1, 0].hist(
    df["petal.length"],
    bins=10,
    color="orange",
    edgecolor="black"
)
axes[1, 0].set_title("Petal Length")
axes[1, 0].set_xlabel("Length")
axes[1, 0].set_ylabel("Frequency")

# Petal Width
axes[1, 1].hist(
    df["petal.width"],
    bins=10,
    color="pink",
    edgecolor="black"
)
axes[1, 1].set_title("Petal Width")
axes[1, 1].set_xlabel("Width")
axes[1, 1].set_ylabel("Frequency")

plt.suptitle("Histograms of Iris Dataset Features")
plt.tight_layout()
plt.show()


# ---------------------------------------------------------
# k) Grouped bar chart:
#    average sepal.length and sepal.width for each variety
# ---------------------------------------------------------

average_sepal = df.groupby("variety")[["sepal.length", "sepal.width"]].mean()

x = range(len(average_sepal))
width = 0.35

plt.figure(figsize=(8, 5))

plt.bar(
    [i - width / 2 for i in x],
    average_sepal["sepal.length"],
    width=width,
    label="Average Sepal Length",
    color="steelblue"
)

plt.bar(
    [i + width / 2 for i in x],
    average_sepal["sepal.width"],
    width=width,
    label="Average Sepal Width",
    color="orange"
)

plt.xticks(list(x), average_sepal.index)
plt.xlabel("Iris Variety")
plt.ylabel("Average Value")
plt.title("Average Sepal Length and Sepal Width")
plt.legend()
plt.show()


# ---------------------------------------------------------
# l) Bubble scatter plot:
#    petal.length vs petal.width
#    bubble size represents sepal.length
# ---------------------------------------------------------

plt.figure(figsize=(8, 6))

for variety, color in colors.items():
    data = df[df["variety"] == variety]

    plt.scatter(
        data["petal.length"],
        data["petal.width"],
        s=data["sepal.length"] * 30,
        color=color,
        alpha=0.6,
        label=variety,
        edgecolors="black"
    )

plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.title("Bubble Plot: Petal Length vs Petal Width")
plt.legend()
plt.grid(True)
plt.show()
```
